**Label Encoding**

Convert emotion string labels to numbers so models can use them.

```python
unique_emotions = df['emotion'].unique()
unique_emotions
```

```python
emotion_numbers = {}
i = 0
for emo in unique_emotions:
    emotion_numbers[emo] = i
    i += 1

emotion_numbers
```

```python
df['emotion'] = df['emotion'].map(emotion_numbers)
```

```python
df.head()
```

**Lowercase Text**

Convert all text to lowercase to ensure uniformity.

```python
df['text'] = df['text'].apply(lambda x: x.lower())
```

```python
df['text'].head()
```

**Remove Punctuation**

Punctuation does not carry useful meaning for emotion classification.

```python
import string

def remove_punc(txt):
    return txt.translate(str.maketrans('', '', string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

```python
df['text'].head()
```

**Remove Numbers**

Numbers usually do not help in emotion detection.

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new
```

```python
df['text'] = df['text'].apply(remove_numbers)
```

```python
df['text'].head()
```

**Remove Emojis**

Keep only ASCII characters to strip out emojis and special unicode symbols.

```python
def remove_emojis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new
```

```python
df['text'] = df['text'].apply(remove_emojis)
```

```python
df['text'].head()
```

**Remove Stopwords**

Stopwords are common words like "the", "is", "in" that do not carry meaningful information.

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
stop_words = set(stopwords.words('english'))
```

```python
df.loc[1]['text']
```

```python
def remove_stopwords(txt):
    words = txt.split()
    cleaned = []
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
    return ' '.join(cleaned)
```

```python
df['text'] = df['text'].apply(remove_stopwords)
```

```python
df.loc[1]['text']
```
