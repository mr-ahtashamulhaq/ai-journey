### **1. Imports and Dataset Loading**

Import the essential data manipulation and visualization libraries. Load the placement dataset, which contains students' CGPA and their corresponding package

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

import warnings
warnings.filterwarnings('ignore')
```

```python
# Load the dataset
df = pd.read_csv('docs/Lecture-035-placement.csv')

# Display structural information and the first few rows
df.info()
df.head()
```

### **2. Exploratory Data Analysis (EDA)**

Visualize the relationship between the independent variable (`cgpa`) and the dependent variable (`package`) using a scatter plot[cite: 14].

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='cgpa', y='package')
plt.title('CGPA vs. Placement Package')
plt.show()
```

### **3. Train-Test Split**

Isolate the features and the target variable[cite: 14]. Split the data into a training set and a testing set (80/20 ratio) to ensure the model can be evaluated on unseen data[cite: 14].

```python
# Define feature matrix X (2D) and target vector y (1D)
X = df[['cgpa']].copy()
y = df['package'].copy()

# Perform the 80/20 split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

### **4. Model Training**

Initialize the `LinearRegression` algorithm and train it using the training split[cite: 14].

```python
lr = LinearRegression()
lr.fit(X_train, y_train)

# Predict the packages for the test set
y_pred = lr.predict(X_test)
```

### **5. Regression Visualization**

Visualize the best-fit line calculated by the model[cite: 14]. The `sns.regplot` function plots the scatter data alongside the linear regression line[cite: 14].

```python
plt.figure(figsize=(8, 5))
sns.regplot(
    x=X_train['cgpa'], 
    y=y_train, 
    scatter_kws={"color": "blue"},  # Training data points
    line_kws={"color": "red"}       # Best-fit regression line
)
plt.title('Best-Fit Regression Line on Training Data')
plt.show()
```

### **6. The Linear Regression Equation ($y = mx + b$)**

A simple linear regression model learns two parameters: the slope ($m$) and the y-intercept ($b$)[cite: 14]. We can extract these directly from the trained model to manually calculate a prediction[cite: 14].

```python
# Extract slope (m) and intercept (b)
m = lr.coef_[0]
b = lr.intercept_

print(f"Slope (m): {m:.4f}")
print(f"y-intercept (b): {b:.4f}")
```

```python
# Manual prediction for a hypothetical CGPA of 9.9
hypothetical_cgpa = 9.9
predicted_package = (m * hypothetical_cgpa) + b

print(f"Predicted package for a CGPA of {hypothetical_cgpa}: {predicted_package:.4f}")
```

### **7. Key Revision Notes**

- **Algorithm Goal:** Simple Linear Regression models the linear relationship between a single independent variable and a continuous dependent variable[cite: 14].
- **Model Parameters:** The model learns by calculating the optimal values for the slope (`lr.coef_`) and the y-intercept (`lr.intercept_`)[cite: 14].
- **Data Shapes:** The feature matrix (`X`) must be 2-dimensional (e.g., a DataFrame or reshaped array), while the target (`y`) is 1-dimensional (e.g., a Series)[cite: 14].
- **Visualization:** `sns.regplot` is an excellent tool for visualizing simple linear relationships as it plots both the data points and the regression line simultaneously[cite: 14].
