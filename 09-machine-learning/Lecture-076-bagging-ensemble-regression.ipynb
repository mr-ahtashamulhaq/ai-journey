## **1. Dataset and Split**

The original Boston dataset loader was removed from scikit-learn. The maintained Diabetes dataset is used as a self-contained replacement for the regression example.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

RANDOM_STATE = 42
sns.set_theme(style="whitegrid")
import warnings
warnings.filterwarnings("ignore")

from sklearn.datasets import load_diabetes
```

```python
diabetes = load_diabetes(as_frame=True)
X = diabetes.data
y = diabetes.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE)
```

## **2. Baseline and Bagging Regressor**

The baseline is a single tree. Bagging fits many trees on bootstrap samples and averages their numeric predictions.

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import BaggingRegressor

baseline = DecisionTreeRegressor(max_depth=6, random_state=RANDOM_STATE)
bagging = BaggingRegressor(
    estimator=DecisionTreeRegressor(max_depth=6, random_state=RANDOM_STATE),
    n_estimators=100, max_samples=0.8, bootstrap=True,
    random_state=RANDOM_STATE, n_jobs=-1
)
models = {"single tree": baseline, "bagging": bagging}
```

```python
results = []
for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    results.append({
        "model": name,
        "MAE": mean_absolute_error(y_test, pred),
        "RMSE": mean_squared_error(y_test, pred) ** 0.5,
        "R2": r2_score(y_test, pred),
    })
pd.DataFrame(results)
```

### **Key Revision Notes**

- Bagging regression averages continuous predictions.
- The ensemble usually reduces variance compared with one high-variance tree.
- `estimator=` is the current scikit-learn parameter name.
- Compare ensembles on unseen data; a higher training score alone is not evidence of improvement.
