### **1. Imports**

Import standard data manipulation, visualization, and modeling libraries. We utilize `make_regression` from scikit-learn to generate synthetic data for this demonstration.

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import warnings

from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Suppress warnings for cleaner output
warnings.filterwarnings('ignore')
```

### **2. Synthetic Data Generation**

Create a dataset with 100 samples and 2 informative features. A noise factor of 50 is introduced to simulate real-world data variance.

```python
# Generate synthetic regression data
X, y = make_regression(n_samples=100, n_features=2, n_informative=2, n_targets=1, noise=50)

# Structure into a pandas DataFrame
df = pd.DataFrame({'feature1': X[:, 0], 'feature2': X[:, 1], 'target': y})

print("Dataset Shape:", df.shape)
df.head()
```

### **3. 3D Data Visualization**

Multiple Linear Regression with two features operates in a 3-dimensional space. We use Plotly to visualize the relationship between the features and the target variable.

```python
# Interactive 3D Scatter Plot
fig = px.scatter_3d(
    df, 
    x='feature1', 
    y='feature2', 
    z='target', 
    title='3D Visualization of Multiple Linear Regression Data'
)

fig.update_traces(marker=dict(size=5, line=dict(width=1, color='DarkSlateGrey')))
fig.show()
```

### **4. Train-Test Split**

Isolate the feature matrix `X` and target array `y`, then split the data using an 80/20 ratio. We use explicit `.copy()` calls to protect the original dataframe.

```python
# Isolate features and target
X_features = df[['feature1', 'feature2']].copy()
y_target = df['target'].copy()

# 80/20 train-test split
X_train, X_test, y_train, y_test = train_test_split(X_features, y_target, test_size=0.2, random_state=3)
```

### **5. Model Training**

Initialize and train the `LinearRegression` model using the training data.

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
```

### **6. Model Evaluation**

Generate predictions against the test set and evaluate the model's performance using standard regression metrics: MAE, MSE, and $R^2$ Score.

```python
# Make predictions
y_pred = lr.predict(X_test)

# Calculate and print metrics
print(f"Mean Absolute Error (MAE): {mean_absolute_error(y_test, y_pred):.4f}")
print(f"Mean Squared Error (MSE):  {mean_squared_error(y_test, y_pred):.4f}")
print(f"R2 Score:                  {r2_score(y_test, y_pred):.4f}")
```

## **7. Key Revision Notes**

- **Multiple Linear Regression vs. Simple:** Simple linear regression models the relationship between one feature and the target with a line ($y = mx + b$). Multiple linear regression models the relationship between *two or more* features and the target, fitting a hyperplane (for 2 features) or higher-dimensional plane in space.
- **`make_regression`:** A utility function from scikit-learn used to generate random regression problems. Highly useful for testing algorithms. Key parameters include `n_samples`, `n_features`, and `noise`.
- **Visualization:** Plotly's `scatter_3d` allows for interactive examination of the multidimensional relationship between two input features and one continuous output.
- **Evaluation interpretation:** The $R^2$ score represents the proportion of the variance in the target variable that is predictable from the independent variables. Higher is better, maxing out at 1.0.
