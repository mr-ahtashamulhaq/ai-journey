### **1. Imports**

Import standard tools for data manipulation and splitting. We do not import `LinearRegression` from `scikit-learn` here because we will build it entirely from scratch

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

import warnings
warnings.filterwarnings('ignore')
```

### **2. Mathematical Formulation & Class Definition**

To build a Simple Linear Regression model from scratch, we use the Ordinary Least Squares (OLS) closed-form formulas to calculate the optimal slope ($m$) and y-intercept ($b$)

1. **Slope ($m$):**  
   $$m = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sum (x_i - \bar{x})^2}$$

2. **Intercept ($b$):**  
   $$b = \bar{y} - m\bar{x}$$

Below is the custom implementation of these mathematical steps in a Python class

```python
class CustomLinearRegression:
    
    def __init__(self):
        self.m = None   # Slope
        self.b = None   # y-intercept
        
    def fit(self, X_train, y_train):
        # Calculate the mean of X and y
        x_mean = X_train.mean()
        y_mean = y_train.mean()
        
        num = 0  # Numerator
        den = 0  # Denominator
    
        # Iterate through all training points to compute the OLS formulas
        for i in range(X_train.shape[0]): 
            num += (X_train[i] - x_mean) * (y_train[i] - y_mean)
            den += (X_train[i] - x_mean) ** 2

        # Assign calculated parameters
        self.m = num / den
        self.b = y_mean - (self.m * x_mean)
    
    def predict(self, X_test):
        # Apply the linear equation y = mx + b
        return (self.m * X_test) + self.b
```

### **3. Dataset Loading & Feature Separation**

Load the placement dataset and extract the independent variable (`cgpa`) and dependent variable (`package`) as flat 1D numpy arrays (via `.values`), as our custom class relies on scalar indexing

```python
# Load dataset
df = pd.read_csv('docs/Lecture-036-placement.csv')

# Extract features as 1D arrays
X = df.iloc[:, 0].values  # CGPA
y = df.iloc[:, 1].values  # Placement Package

print("X shape:", X.shape)
print("y shape:", y.shape)
```

### **4. Train-Test Split**

Split the data into training and testing sets to evaluate our custom algorithm on unseen data

```python
# 80/20 Train-Test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)

print("Training samples:", X_train.shape[0])
```

### **5. Model Training & Parameter Extraction**

Instantiate the custom class and call the `.fit()` method. This will calculate the slope and intercept over the 160 training examples

```python
# Initialize and train custom model
lr = CustomLinearRegression()
lr.fit(X_train, y_train)

# Display learned parameters
print(f"Learned Slope (m): {lr.m:.4f}")
print(f"Learned Intercept (b): {lr.b:.4f}")
```

### **6. Making Predictions**

Use the `.predict()` method to generate a prediction for a specific value from the test set

```python
# Predict the package for the first student in the test set
test_cgpa = X_test[0]
predicted_package = lr.predict(test_cgpa)

print(f"CGPA input: {test_cgpa}")
print(f"Predicted Package: {predicted_package:.4f}")
```

### **7. Key Revision Notes**

- **Under the Hood:** Simple Linear Regression is an optimization problem solved analytically using the Ordinary Least Squares (OLS) approach. It finds the line of best fit by minimizing the sum of squared residuals
- **Mathematical Foundation:** The slope represents the covariance of X and y divided by the variance of X. The intercept is simply shifting the line so it passes exactly through the mean coordinates $(\bar{x}, \bar{y})$
- **API Consistency:** We implemented `fit()` and `predict()` methods to mirror `scikit-learn`'s object-oriented structure, demonstrating how machine learning estimators are constructed functionally
