## **Column Transformer**

1. Applies different preprocessing techniques to different columns in one step.
   
2. Useful when a dataset has mixed feature types (numerical + categorical).


3. Fits all transformers on the training data and applies the same transformations to the test data.

```python
import numpy as np
import pandas as pd
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('docs/Lecture-015-covid_toy.csv')
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],
                                                test_size=0.2)
```

## **Column Transformer**

`'tnf1'` → Name of this transformer (can be any name).

`StandardScaler()` → Transformation to apply.

`['age', 'salary']` → Columns on which the transformation is applied.


`remainder ='passthrough'` --> Keeps all remaining columns unchanged.

`remainder='drop'` --> Drops all columns that are not explicitly listed.


```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test).shape
```

`fit_transform()` → Learn preprocessing parameters from training data and transform it.

`transform()` → Apply the same learned transformations to the test data as well.
