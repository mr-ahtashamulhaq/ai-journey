### **Environment Setup and Data Loading**

Import standard machine learning models and metrics, then load the Diabetes dataset to evaluate different regularization techniques.

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X, y = load_diabetes(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

### **Baseline: Linear Regression**

Train a standard Ordinary Least Squares (OLS) Linear Regression model to establish a baseline $R^2$ score.

```python
# Linear Regression Baseline
lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)
y_pred_lin = lin_reg.predict(X_test)

print(f"Linear Regression R2: {r2_score(y_test, y_pred_lin):.4f}")
```

### **Ridge and Lasso Regression Comparison**

Apply Ridge (L2 penalty) and Lasso (L1 penalty) to observe how independent regularization affects the model's performance.

```python
# Ridge Regression
ridge_reg = Ridge(alpha=0.1)
ridge_reg.fit(X_train, y_train)
y_pred_ridge = ridge_reg.predict(X_test)
print(f"Ridge R2: {r2_score(y_test, y_pred_ridge):.4f}")

# Lasso Regression
lasso_reg = Lasso(alpha=0.01)
lasso_reg.fit(X_train, y_train)
y_pred_lasso = lasso_reg.predict(X_test)
print(f"Lasso R2: {r2_score(y_test, y_pred_lasso):.4f}")
```

### **Elastic Net Regression Implementation**

Elastic Net combines both L1 and L2 penalties. The `l1_ratio` parameter dictates the mix between Lasso and Ridge. In this implementation, `l1_ratio=0.9` signifies a model heavily weighted toward Lasso (90%) but retaining a small Ridge penalty (10%) for stability.

```python
# ElasticNet Regression
elastic_reg = ElasticNet(alpha=0.005, l1_ratio=0.9)
elastic_reg.fit(X_train, y_train)
y_pred_elastic = elastic_reg.predict(X_test)
print(f"ElasticNet R2: {r2_score(y_test, y_pred_elastic):.4f}")
```

### **Key Revision Notes**

- **Elastic Net Mechanism:** It overcomes the limitations of Lasso (which can randomly select one feature from a group of highly correlated features) by adding a Ridge penalty, which encourages grouped selection.
- **Hyperparameters:** 
  - `alpha`: Controls the overall strength of the combined penalty.
  - `l1_ratio`: Controls the balance. A ratio of 1.0 is pure Lasso, while 0.0 is pure Ridge.
- **Performance Observation:** By fine-tuning both `alpha` (0.005) and `l1_ratio` (0.9), Elastic Net can outperform both isolated Ridge and Lasso models.
