### **Environment Setup and Data Loading**

Import evaluation metrics and prepare the dataset independently so this notebook can run seamlessly.

```python
import numpy as np
import pandas as pd 
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score
```

```python
df_iris = pd.read_csv('docs/Lecture-057-Iris.csv')

encoder = LabelEncoder()
df_iris['Species'] = encoder.fit_transform(df_iris['Species'])
```

### **Data Splitting and Model Training**

Prepare the data splits and train models to evaluate advanced metrics.

```python
X = df_iris.iloc[:, 0:-1]
y = df_iris.iloc[:, -1]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
log_reg_model = LogisticRegression(max_iter=200)
tree_model = DecisionTreeClassifier()

log_reg_model.fit(X_train, y_train)
tree_model.fit(X_train, y_train)

y_pred_log_reg = log_reg_model.predict(X_test)
y_pred_tree = tree_model.predict(X_test)
```

### **Advanced Metrics Evaluation**

Calculate Precision, Recall, and the F1 Score for both models. For multi-class problems, setting `average=None` returns the respective metric for each class individually.

```python
print("For Logistic Regression Model")
print("-" * 50)
print("Precision (per class):", precision_score(y_test, y_pred_log_reg, average=None))
print("Recall (per class):   ", recall_score(y_test, y_pred_log_reg, average=None))
print("F1 Score (per class): ", f1_score(y_test, y_pred_log_reg, average=None))
```

```python
print("For Decision Tree Model")
print("-" * 50)
print("Precision (per class):", precision_score(y_test, y_pred_tree, average=None))
print("Recall (per class):   ", recall_score(y_test, y_pred_tree, average=None))
print("F1 Score (per class): ", f1_score(y_test, y_pred_tree, average=None))
```

### **Key Revision Notes**

- **Precision:** Answers *"Out of all instances the model predicted as positive, how many were actually positive?"* Useful when the cost of a False Positive is high.
- **Recall:** Answers *"Out of all actual positive instances, how many did the model correctly identify?"* Useful when the cost of a False Negative is high.
- **F1 Score:** The harmonic mean of Precision and Recall. Strongly penalizes extreme values, making it highly effective for evaluating models on imbalanced datasets.
- **Multi-class `average` Parameter:** Using `average=None` is essential in non-binary datasets to diagnose performance per specific category rather than masking flaws in a macro/micro average.
