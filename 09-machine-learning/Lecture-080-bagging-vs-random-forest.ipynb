## **Bagging versus Random Forest**

### **Topic Roadmap**

**1. Create a common classification dataset**

**2. Fit bagging with trees**

**3. Fit a random forest**

**4. Compare structure and performance**

**5. Key revision notes**

## **1. Common Dataset**

Both models are evaluated on identical data. This isolates the algorithmic difference: ordinary bagging can use any estimator, while a random forest also randomizes the feature subset at each tree split.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier, RandomForestClassifier

RANDOM_STATE = 42
X, y = make_classification(n_samples=2000, n_features=10, n_informative=4, n_redundant=1, random_state=RANDOM_STATE)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=RANDOM_STATE)
```

## **2. Fit Both Ensembles**

The bagging model uses full feature vectors for each tree. The forest uses random feature subsets through `max_features`.

```python
bagging = BaggingClassifier(
    estimator=DecisionTreeClassifier(random_state=RANDOM_STATE),
    n_estimators=150, max_samples=0.8, max_features=1.0,
    bootstrap=True, random_state=RANDOM_STATE, n_jobs=-1
)
forest = RandomForestClassifier(
    n_estimators=150, max_features="sqrt", random_state=RANDOM_STATE, n_jobs=-1
)
for name, model in [("Bagging", bagging), ("Random forest", forest)]:
    model.fit(X_train, y_train)
    print(name, f"accuracy = {model.score(X_test, y_test):.3f}")
```

## **3. Structural Comparison**

The stored feature indices make the difference visible. Bagging with `max_features=1.0` uses all features for every base estimator; the forest varies feature candidates internally at each split.

```python
bagging.fit(X_train, y_train)
print("Bagging feature count for first estimator:", bagging.estimators_features_[0].size)
print("Total input feature count:", X.shape[1])
print("Random forest max_features:", forest.max_features)
```

### **Key Revision Notes**

- Bagging means bootstrap aggregation of a chosen base estimator.
- A random forest is specialized bagging of decision trees with random feature selection.
- Random feature selection reduces correlation between trees.
- The best choice depends on estimator type, feature correlation, data size, and tuning.
