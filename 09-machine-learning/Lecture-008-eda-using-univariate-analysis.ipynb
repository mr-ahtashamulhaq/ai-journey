## **EDA - Univariate Analysis**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('docs/Lecture-008-train.csv')
```

```python
df.head()
```

# 1. Categorical Data

## a. Countplot

```python
sns.countplot(data=df, x='Survived')
```

## b. PieChart

```python
df['Sex'].value_counts().plot(kind='pie' , autopct='%.2f')
```

# 2. Numerical Data

## a. Histogram

```python
sns.displot(data=df, x="Age", kind="hist")
```

## b. Distplot

```python
sns.displot(data=df , x='Age' , kind='kde', fill=True)
```

## c. BoxPlot

```python
sns.boxplot(data=df, x='Age')
```
