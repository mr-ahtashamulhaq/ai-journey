### **1. Imports & Dataset Preparation**

Import the required libraries, load the diabetes dataset, and split it into training and testing sets[cite: 3].

```python
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split

# Load diabetes dataset
X, y = load_diabetes(return_X_y=True)

# Create training and testing splits
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

### **2. OLS Baseline Model**

Train a standard Ordinary Least Squares (OLS) Linear Regression model to establish a performance baseline[cite: 3].

```python
# Initialize and train the baseline model
reg = LinearRegression()
reg.fit(X_train, y_train)

# Evaluate baseline performance
y_pred_ols = reg.predict(X_test)
baseline_r2 = r2_score(y_test, y_pred_ols)

print(f"Baseline OLS R2 Score: {baseline_r2:.4f}")
```

### **3. Batch Gradient Descent Implementation**

Build a custom Batch Gradient Descent regressor from scratch.

This class iteratively updates the coefficients and intercept by calculating the loss gradients across the entire training dataset in a single matrix operation[cite: 3].

```python
class GDRegressor:
    def __init__(self, learning_rate=0.01, epochs=100):
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs
        
    def fit(self, X_train, y_train):
        # Initialize intercept to 0 and coefficients to an array of ones
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])

        # Iteratively update weights over the specified number of epochs
        for i in range(self.epochs):
            # Calculate predictions for the entire batch
            y_hat = np.dot(X_train, self.coef_) + self.intercept_
            
            # Calculate loss gradients with respect to intercept and coefficients
            intercept_der = -2 * np.mean(y_train - y_hat)
            coef_der = -2 * np.dot((y_train - y_hat), X_train) / X_train.shape[0]

            # Apply parameter updates
            self.intercept_ = self.intercept_ - (self.lr * intercept_der)
            self.coef_ = self.coef_ - (self.lr * coef_der)
            
    def predict(self, X_test):
        # Calculate final predictions using the learned parameters
        return np.dot(X_test, self.coef_) + self.intercept_
```

### **4. Model Evaluation**

Train the custom Batch Gradient Descent model and evaluate its performance[cite: 3].

```python
# Initialize the custom model with tuned hyperparameters
gdr = GDRegressor(epochs=400, learning_rate=0.5)

# Train the model
gdr.fit(X_train, y_train)

# Evaluate custom model performance
y_pred_gd = gdr.predict(X_test)
gd_r2 = r2_score(y_test, y_pred_gd)

print(f"Batch Gradient Descent R2 Score: {gd_r2:.4f}")
```

### **Key Revision Notes**

- **Batch Gradient Descent:** Evaluates the gradient of the cost function against the *entire* dataset to perform a single parameter update per epoch[cite: 3].
- **Vectorization:** Achieves computational efficiency by utilizing dot products (`np.dot`) to perform batch predictions and derivative calculations without explicit loops[cite: 3].
- **Hyperparameter Tuning:** Performance heavily depends on `learning_rate` and `epochs`. The objective is to find a balance where the loss converges smoothly without oscillating or taking too long.
