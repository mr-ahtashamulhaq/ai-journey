### **1. Imports & Setup**

Import essential libraries for data generation, modeling, and evaluation.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

### **2. Dataset Preparation**

Generate a synthetic regression dataset and split it into training and testing sets.

The training set is used to learn patterns, while the testing set evaluates model performance on unseen data.

```python
# Generate synthetic linear regression data
X, y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1, noise=20, random_state=13)

# Create training and testing splits
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

### **3. OLS Baseline Model**

Train a standard Ordinary Least Squares (OLS) Linear Regression model to establish a performance baseline.

```python
# Initialize and train the baseline model
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

# Evaluate baseline performance
y_pred_lr = lr_model.predict(X_test)
baseline_r2 = r2_score(y_test, y_pred_lr)

print(f"Baseline OLS R2 Score: {baseline_r2:.4f}")
```

### **4. Gradient Descent Implementation**

Build a custom Gradient Descent regressor from scratch.

This class iteratively updates the slope (`m`) and intercept (`b`) using the gradients of the Mean Squared Error (MSE) loss function to find the optimal parameters.

```python
class GDRegressor:
    def __init__(self, learning_rate=0.001, epochs=50):
        self.m = 100   # Initial weight guess
        self.b = -120  # Initial bias guess
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self, X, y):
        # Iteratively update m and b to minimize loss
        for i in range(self.epochs):
            # Calculate loss gradients with respect to b and m
            loss_slope_b = -2 * np.sum(y - self.m * X.ravel() - self.b)
            loss_slope_m = -2 * np.sum((y - self.m * X.ravel() - self.b) * X.ravel())

            # Apply parameter updates
            self.b = self.b - (self.lr * loss_slope_b)
            self.m = self.m - (self.lr * loss_slope_m)
        
    def predict(self, X):
        # Calculate predictions using the learned parameters
        return self.m * X + self.b
```

### **5. Model Evaluation**

Train the custom Gradient Descent model and compare its performance against the baseline.

```python
# Initialize and train the custom Gradient Descent model
gd_model = GDRegressor(learning_rate=0.001, epochs=50)
gd_model.fit(X_train, y_train)

# Evaluate custom model performance
y_pred_gd = gd_model.predict(X_test)
gd_r2 = r2_score(y_test, y_pred_gd)

print(f"Custom Gradient Descent R2 Score: {gd_r2:.4f}")
```

### **Key Revision Notes**

- **Gradient Descent:** An optimization algorithm that iteratively adjusts model parameters (weights and biases) to minimize a predefined loss function.
- **Learning Rate:** Controls the magnitude of parameter updates. A learning rate that is too large causes divergence, while one that is too small results in slow convergence.
- **Epochs:** The total number of complete passes the algorithm makes through the entire training dataset.
- **Gradients:** The partial derivatives of the loss function. They point in the direction of steepest ascent; updating parameters in the opposite direction continuously reduces the error.
